### 1.7 Сколько карточек содержат список пар в Значение Инцидента (группировка)?

Всего:

```python
df['Значение Инцидента (группировка)'].str.contains(r'\|', na=False).mean()
```

непустых:

```python
col = df['Значение Инцидента (группировка)']
has_value = col.notna()
has_pairs = col.str.contains(r'\|', na=False)

(has_pairs & has_value).sum() / has_value.sum()
```
